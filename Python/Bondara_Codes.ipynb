```python
date_columns = ['LoanDate','ContractEndDate','FirstPaymentDate','DefaultDate']
for i in date_columns:
    df[i]=pd.to_datetime(df[i],errors='coerce')
```

```python
df.dtypes
```

```python
numeric_columns = ['MonthlyPayment','ExpectedLoss','ExpectedReturn','ProbabilityOfDefault','DebtToIncome']
for i in numeric_columns:
    df[i].fillna(df[i].median(),inplace=True)
```

```python
categorical_columns = ['Gender','Education','EmploymentStatus','OccupationArea']
for i in categorical_columns:
    df[i].fillna(df[i].mode()[0],inplace=True)
```

```python
df['Rating'] = df['Rating'].fillna('Unkown')
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
df['DefaultFlag'] = df["DefaultDate"].notna().astype(int)
```

```python
df['OutstandingAmount'] = df['Amount'] - df['PrincipalPaymentsMade']
```

```python
df['RepaymentPercentage'] = (df['PrincipalPaymentsMade']/df['Amount'])*100
```

```python
df['TotalRecived'] = (df['PrincipalPaymentsMade']+df['InterestAndPenaltyPaymentsMade'])
```

```python
Today = pd.Timestamp.today()
df['LoanAgeDays'] = (Today-df['LoanDate']).dt.days
```

```python
df["RiskCategory"] = pd.cut(
    df["ProbabilityOfDefault"],
    bins=[0, 0.20, 0.40, 1],
    labels=["Low", "Medium", "High"],
    include_lowest=True
)
```

```python
df["NetROI"] = (
    df["ExpectedReturn"] -
    df["ExpectedLoss"]
)
```
